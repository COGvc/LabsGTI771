## Introduction

Ce deuxième laboratoire porte sur la définition et l’extraction de primitives artisanales (*handcrafted*, *engineered*) sur des visages. On reprendre le problème [Facial Expression Recognition (FER)](https://www.kaggle.com/c/challenges-in-representation-learning-facial-expression-recognition-challenge/data), dont le but est de classer des visages dans sept (7) catégories d'émotions.

Dans le contexte de l'apprentissage automatique, les primitives artisanales (*engineered*) désignent des caractéristiques ou des représentations de données conçues manuellement par des experts du domaine. Ces primitives sont souvent le fruit d'une analyse approfondie des données et d'une compréhension solide des mécanismes sous-jacents du problème à résoudre. Elles servent de point de départ pour entraîner des modèles d'apprentissage automatique.

Les caractéristiques globales, également appelées caractéristiques holistiques, analysent l'ensemble du visage. Elles saisissent les propriétés structurelles et géométriques générales du visage, comme sa forme et la position relative des points de repère faciaux.

#### Méthodes d'extraction de caractéristiques globales
- <u>Caractéristiques géométriques</u> : Cette approche se concentre sur les relations spatiales et les distances entre les principaux points de repère faciaux (par exemple, les coins des yeux et de la bouche, le bout du nez). Ces mesures sont ensuite utilisées comme caractéristiques. Bien que simple et efficace en termes de calcul, cette méthode peut être sensible aux mouvements de la tête et aux petites imprécisions de la détection des points de repère.

- <u>Scale-Invariant Feature Transform (SIFT)</u> et <u>Speeded Up Robust Features (SURF)</u> : Bien que souvent utilisées pour la reconnaissance d'objets, ces méthodes peuvent être appliquées globalement pour extraire des caractéristiques de l'image faciale entière. Elles identifient des points clés et leurs descripteurs, qui sont insensibles aux changements d'échelle, de rotation et d'éclairage.

- <u>Principal Component Analysis (PCA)</u> : En tant que technique de réduction de la dimensionnalité, l'ACP peut être utilisée comme extracteur de caractéristiques globales. Elle identifie les composantes principales (vecteurs propres) de l'ensemble de données d'images faciales, souvent appelées « visages propres » (eigenfaces). Une expression peut alors être représentée comme un vecteur de poids basé sur ces visages propres. Cette méthode capture la structure globale du visage et ses variations.

- <u>Active Appearance Models (AAM)</u> : Les AAM sont des modèles statistiques qui combinent un modèle de forme faciale avec un modèle de texture faciale. Le modèle s'adapte à une nouvelle image de visage, et les paramètres nécessaires pour cet ajustement deviennent les caractéristiques. Ces paramètres encodent à la fois la déformation de la forme globale (caractéristiques géométriques) et les changements de texture (caractéristiques d'apparence) du visage, ce qui en fait une méthode hybride très puissante.

- <u>Triangulation géométrique</u> : Cette méthode consiste à placer des points de repère sur les points faciaux clés (comme les coins des yeux et de la bouche, et le bout du nez), puis à calculer les caractéristiques basées sur les relations géométriques entre eux. Ces caractéristiques peuvent inclure les distances, les angles ou les aires des triangles formés par ces points. Par exemple, la surface du triangle formé par les deux yeux et le bout du nez peut changer lors d'expressions comme la surprise ou la colère.

## 1h: Extraction de primitives locales

Entrées: *data-cleaned-pre* et *data-cleaned-pre-reeq*<br> 
Sorties: fichier avec les vecteurs de caractéristiques 
- *data-cleaned-pre-train-###.csv*, *data-cleaned-pre-reeq-train-###.csv* et *data-cleaned-pre-test-###.csv* <br>
où *###* désigne l'algorithme d'extraction de primitives choisis.

Obs: Utilisez le format `hdf5` (*.h5*) s'il s'agit d'une dimension élevée.

Les caractéristiques locales sont des descripteurs qui capturent des détails fins de régions spécifiques du visage, telles que les yeux, les sourcils, la bouche et le nez. Ces régions sont particulièrement importantes pour exprimer des émotions. Les méthodes locales sont robustes face aux variations d'éclairage, de pose et aux occlusions, car elles se concentrent sur de petites zones localisées de l'image plutôt que sur l'ensemble du visage. [Koutlas et Fotiadis (2008)](http://dx.doi.org/10.4018/9781605663142.ch016) suggèrent que certaines régions du visage, notamment autour des points de référence clés, contiennent des informations plus discriminantes pour la reconnaissance d'expressions faciales. Ils ont proposé un ensemble de 20 points de référence.

Afin de mieux capturer ces informations locales, une approche courante consiste à diviser chaque image en $n$ zones de taille égale (voir la figure ci-bas). Pour chaque zone $z_i$ (avec $i$ allant de $0$ à $n−1$), nous appliquerons l'algorithme d'extraction de caractéristiques sélectionné, ce qui générera un vecteur de caractéristiques ($d$ dimensions) pour chaque zone. Les $n$ vecteurs de caractéristiques obtenus pour chaque zone seront ensuite concaténés pour former un descripteur global plus riche en information ($n\times d$ dimensions).

![Exemples de FER](https://b2633864.smushcdn.com/2633864/wp-content/uploads/2021/04/face_reco_lbps_header.png?lossy=1&strip=1&webp=1)

Cette approche permet de:

- Mieux représenter les variations locales: En divisant l'image en zones, nous sommes plus sensibles aux variations locales d'intensité, de texture ou de forme, qui peuvent être cruciales pour différencier les expressions faciales.
- Intégrer des connaissances a priori: Bien que nous ne connaissions pas la localisation exacte des points de référence proposés par Koutlas et Fotiadis, cette approche nous permet d'explorer l'espace facial de manière plus fine.

**Remarque**: Le choix du nombre de zones $n$ est un hyperparamètre à ajuster en fonction de la complexité des expressions à reconnaître et de la résolution des images.

**Visualisation**: Il serait intéressant de visualiser les zones choisies sur quelques exemples d'images pour s'assurer qu'elles correspondent bien aux régions d'intérêt.

#### Méthodes d'extraction de caractéristiques locales
<u>Histogramme de Gradients Orientés (HOG)</u> : Cette méthode décrit la forme et l'apparence des objets dans une image en comptant les occurrences d'orientation des gradients dans des portions localisées. Pour la reconnaissance des émotions, les caractéristiques HOG peuvent saisir la direction et l'amplitude des changements d'intensité de pixels qui correspondent au mouvement des muscles faciaux.

<u>Local Binary Patterns (LBP)</u> : Le LBP est un descripteur de texture qui attribue un code binaire à chaque pixel en fonction des valeurs d'intensité de ses voisins. Il est très efficace pour capturer la texture locale et les micro-motifs, qui sont cruciaux pour détecter les changements subtils dans les expressions faciales. Des variantes comme LBP-TOP (LBP sur trois plans orthogonaux) étendent ce concept aux vidéos, capturant à la fois des informations spatiales et temporelles.

<u>Weber Local Descriptor (WLD)</u> : Ce descripteur est robuste face aux variations d'éclairage, de contraste et de configurations géométriques. Il se compose de deux éléments : l'excitation différentielle et l'orientation. L'excitation différentielle décrit la différence d'intensité entre un pixel et ses voisins, tandis que l'orientation capture des informations de gradient.

<u>Filtres de Gabor</u> : Les filtres de Gabor sont largement utilisés pour l'analyse de texture et peuvent capturer efficacement des caractéristiques locales telles que les bords, les lignes et les textures à différentes échelles et orientations. Ils sont particulièrement utiles pour représenter les détails faciaux qui contribuent aux expressions.

<u>FAST (Features from Accelerated Segment Test)</u> et <u>Harris Corner Detection</u>: Ces méthodes sont principalement utilisées pour détecter les points d'intérêt ou les coins dans une image. Bien qu'elles ne soient pas directement des descripteurs d'émotions, les emplacements et les caractéristiques de ces points angulaires sur un visage peuvent fournir des informations structurelles locales précieuses qui changent avec les expressions.
